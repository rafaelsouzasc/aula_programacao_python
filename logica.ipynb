```python
#Realizando soma de variáveis diferentes
q_seguranca = 5
s_seguranca = 3000

q_docente = 16
s_docente = 6000

q_diretoria = 1
s_diretoria = 12500

q_trabalhadores = sum([q_diretoria,q_docente,q_seguranca])
print(q_trabalhadores)
```

```python
#Valor do salário mais alto
salario_max= max([s_seguranca,s_docente,s_diretoria])
print(salario_max)
```

```python
#Utilizando um dicionário para trazer o maior valor e o cargo

salarios = {
    'Segurança': s_seguranca,
    'Docente': s_docente,
    'Diretoria': s_diretoria
}

cargo_maior_salario = max(salarios, key=salarios.get)
print(cargo_maior_salario)

print(f"O cargo com maior salário é: {cargo_maior_salario}")
print(f"O valor do salário é: R$ {salarios[cargo_maior_salario]}")

'''
 Ao passar o dicionário salarios para a função max (), o Python vai verificar apenas as chaves, 
 ex: Segurança. Ao utilizar o parâmetro key=salarios.get, ele utiliza a chave e joga dentro da função
 .get().

 Ponto importante: salarios.get (sem os parênteses ()). Isso porque não está sendo executada a função
 ali na hora. Está apenas entregando a função para o max() 

 A função get() ela garante um retorno a sua função caso ela não encontre o valor definido.
 Exemplo: se o parâmetro estivesse escrito apenas key=salarios e não existisse esse dicionário
 ou DataFrame a função quebraria. Com o get garante que ela retorne None e não quebre.

 É possível também passar algum valor caso não seja encontrado.
 Exemplo: cargo_maior_salario = max(salarios,key=salarios.get('Sem valor referente')). 
 Isso garante que caso não ache o valor retorno com a string definida 
'''

print()
#Utilizando estruturas lógicas
if salario_max == s_diretoria:
    nome_maior_salario = "Diretoria"
elif salario_max == s_docente:
    nome_maior_salario = "Docente"
else:
    nome_maior_salario ="Segurança"

print(f"O cargo com maior salário é: {nome_maior_salario}")
```

]

Ao usar apenas vírgulas dentro das chaves, você cria um Set (Conjunto).
Um Set é uma coleção de itens "soltos" e sem ordem definida. Não existe relação entre a palavra 'idade' e o número 45 — eles são apenas três itens diferentes jogados no mesmo saco.

O elif permite adicionar múltiplas condições a um código e é executado somente se as condições do if ou de todos os elifs anteriores forem falsas. 
Assim, o elif fornece uma maneira de verificar várias condições diferentes sem precisar usar vários comandos if. 
No exemplo, o elif é usado para verificar se o valor do imóvel está entre 100000 e 500000 e também se o valor é igual ou superior a 500000, 
mas essa verificação só é feita caso a condição anterior seja falsa.

```python
'''
O continue e o break são comandos de controle que podem ser executados com o for e while.

O continue interrompe a iteração atual do laço e salta para a próxima, ou seja, retorna ao início do código.
'''
for i in range(1,6):
    if i == 2:
        continue
    print(i)

print()

'''
O código imprime todos os números de 1 a 5, exceto 2. Quando o valor de i for 2, o continue salta para a próxima iteração, ignorando o comando print abaixo da condicional na iteração atual
'''

'''
Já o break interrompe a execução do laço completamente, saindo do bloco de código.
'''

for i in range(1,6):
    if i == 4:
        break
    print(i)

'''
O código imprime todos os número de 1 à 6 e quando o valor de i for igual a 4, o break interrompe a execução do laço e sai dele, 
ignorando qualquer outra iteração que esteja dentro da estrutura.
'''
```

Em resumo, o continue permite pular uma iteração específica do laço, enquanto o break permite finalizar o laço e continua a execução dos comandos fora dele.

## Estrutura de dados

- Listas: é uma estrutura de dados que permite armazenar uma coleção de elementos.


```python
lista = ['Fabricio Daniel', 9.5,9.0,8.0, True]
print(lista)
```

```python
#Atribuição a uma nova nota
lista[3] = 10
print(lista)
```

```python
media_fabricio = (lista[1]+lista[2]+lista[3])/3
print(f"{media_fabricio:.2f}")

#Podemos escrever essa média utilizando apenas os valores float da lista de outra forma
valores_media = lista[1:4]
media_1 = sum(valores_media) / len(valores_media)
print(f"Os valores da média é {media_1:.2f}")

#Porém, temos um problema. Caso a lista cresça, a posição do itens pode alterar. Outra maneira de escrever é:
valores_float=[x for x in lista if isinstance(x, int)]
media_2 = sum(valores_float)/ len(valores_float)
print(f"Média atualizada é {media_2:.2f}")

```

```python
serie_numeros = list(range(1,10))
print(serie_numeros)

media_serie_numeros = sum(serie_numeros) / len(serie_numeros)
print(f"{media_serie_numeros:.2f}")

media_serie_numeros_1 = [x for x in serie_numeros if isinstance(x,(int,float))]
media = sum(media_serie_numeros_1)/ len(media_serie_numeros_1)
print(f"{media:.2f}")
```

Mesmo assim é possível transformar uma string em uma lista através do método split(). Esse método corta a string sempre que encontrar o delimitador informado.

```python
duvida = 'Quem veio antes? O ovo ou a galinha?'
lista_de_palavras = duvida.split('?')
print(lista_de_palavras)

#o separado utilizado é '?', mas poderia ser também o espaço ' ', que iria trazer cada palavra como um item.
```

### Questão de aula.



Pelo método append acrescentamos um item ao final da nossa lista. Com o método extend, acrescentamos vários itens ao final da nossa lista. 

Em um caso utilizei a função split, para separa uma string e transforma-la em uma lista. Porém quero acrescentar um ponto a cada item que foi separado dessa string. Como os métodos informado anteriormente não atenderia esse caso, como posso executar essa tarefa ?

```python
frase = "Olá mundo Python"

lista_palavras = frase.split() #Como não passei nenhum delimitador, ele reconhece o espaço.

listas_com_pontos = [] #Criei uma nova lista, assim não altero a lista anterior

for palavra in lista_palavras:
    listas_com_pontos.append(palavra+". ")
print(listas_com_pontos)
```

###  Outras manipulações

Insert

Insere algum elemento em uma determinada posição da lista.
insert(indice, elemento) -> indice é a posição do item inserido na lista e elemento é o que será inserido.

```python
raca_caes = ['Labrador Retriever',

             'Bulldog Francês',

             'Pastor Alemão',

             'Poodle']

raca_caes.insert(1, 'Golden Retriver')
print(raca_caes)
```

Pop

Remove o elemento de uma determinada posição da lista e o retorna como saída da execução do método. 

A diferença do método pop para o remove é que o pop ele remove e retorna o item removido. O método remove, ele remove e retorna a lista sem esse item.

```python
raca_caes.pop(1)
```

```python
raca_caes
```

Index

Retorna o íncide de um elemento específico na lista. 

```python
raca_caes.index('Pastor Alemão')
```

Sort

Organiza os elementos da lista em ordem crescente ou descrescente. Caso seja palavras, a ordem é alfabética.

```python
raca_caes.sort()
print(raca_caes)

#Para fazer em ordem decrescente
raca_caes.sort(reverse=True)
print(raca_caes)
```

1) Faça um programa que tenha a seguinte lista contendo os valores de gastos de uma empresa de papel [2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08]. Com esses valores, faça um programa que calcule a média de gastos. Dica: use as funções built-in sum() e len().

```python
valores = [2172.54, 3701.35, 3518.09, 3456.61, 3249.38, 2840.82, 3891.45, 3075.26, 2317.64, 3219.08]

#Uma maneira de se fazer:
media_gastos = sum(valores)/len(valores)
print(f"Média de valores gastos R${media_gastos: .2f}")

#Segunda maneira de ser feito, utilizando def
def calcular_media(valores):
    if not valores:
        return 0
    return sum(valores)/len(valores)
print(f"Media de valores gastos R${calcular_media(valores): .2f}")
```

5) Faça um programa que, ao inserir um número qualquer, cria uma lista contendo todos os números primos entre 1 e o número digitado.

```python
numero = int(input('Digite um número inteiro aqui: '))

lista_primos = []

for num in range(2,numero):
    primo=True
    for teste_divisiveis in range(2, num):
        if num % teste_divisiveis == 0:
            primo = False
            break
        if primo:
            lista_primos.append(num)
print(f'Lista de números primos: {lista_primos}')
```
